By sampling an initial noise vector, the model commits to either the top or bottom gap. Integrating this velocity network iteratively transforms the noise into valid expert actions without interpolating an unviable middle path, thereby achieving excellent performance.

4. **Explain** in 3–4 sentences: why does DAgger improve over rounds? What role does the deterministic expert play? How does this DAgger's approach solve the challenges mentioned previously for the MSE regression policy?  
    <font color="red">**Include your explanation here.**</font>  
DAgger improves over rounds because its iterative on-policy rollouts allow the agent to visit drifted states and receive direct expert corrections, actively mitigating compounding errors. The deterministic expert eliminates the multimodal ambiguity of the original expert by always picking the upper gap, thus providing conflict-free target actions. This deterministic relabeling converts the two-choice distribution into a single-mode regression problem, which is easily optimized using the MSE objective. By combining on-policy distribution matching with definitive expert targets, DAgger effectively resolves not only distribution shift but also multi-choice averaging and significantly outperforms vanilla BC.
