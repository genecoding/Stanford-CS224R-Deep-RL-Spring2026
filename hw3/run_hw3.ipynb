<font color="red">**(2) Explanation:**</font>  
$\zeta$ = 0.9 performs significantly better because a larger expectile places more weight on the upper bounds of the TD targets present within the dataset, whereas $\zeta$ = 0.2 downweights them. This allows the value function $V_\phi(s)$ to estimate the maximum potential of the data distribution rather than the average. By using an upper expectile, IQL extracts a high-performing policy while completely avoiding the overestimation caused by querying out-of-distribution actions.

<font color="red">**Explanation:**</font>  
IQL performs substantially better on the harder, longer-horizon task because it avoids actor-critic error propagation by completely decoupling policy learning from value estimation. While AWAC constrains the policy toward the behavior policy and uses actor-sampled actions to evaluate the next-state target value, any minor policy OOD slip propagates compounding overestimation errors down the long horizon. In contrast, IQL utilizes an upper expectile ($\zeta$ = 0.9) to directly approximate the maximum state values available within the dataset support, allowing it to accurately stitch optimal trajectories together without ever querying out-of-sample actions during critic updates.
